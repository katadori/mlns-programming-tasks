### 1.1 Visual Studio Code のインストール

- プログラムを書くためのエディタです。
- 便利なプラグインがたくさんあるので使い勝手が良い。
- 公式サイト ([Visual Studio Code](https://code.visualstudio.com/download)) からインストールしてください。

### 1.2 Python のインストール

- 公式サイト（[python.org](https://www.python.org/)）から最新版の Python をインストールしてください。
- インストール後、ターミナルまたはコマンドプロンプトで以下のコマンドを実行し、バージョンが表示されるか確認します。

```bash
python3 --version
# または Windows では
python --version
```

- インストール後、VSCode 拡張機能から「Python」「Python Environments」拡張機能をインストールします。

### 1.3 仮想環境を作る

`C:\Users\"username"\Documents\my-project-1`

- 別にどこでもいいですが、上記のようにどこかにプログラミング用のフォルダ(例えば `my-project-1`)を作ります。
- そしたらいま作ったフォルダを vscode で開きましょう。
- いま開いたフォルダがきちんと今作成したフォルダになっているかを確認しましょう。
- 確認出来たら、そのプロジェクトフォルダ内で以下の拡張機能の一番下（Python のアイコンのやつ、Python Environmentsという拡張機能）をクリックします。
  


- すると以下のような画面になると思います。



- そしたら、venvの「+」ボタンをクリック



- そしたらこんな感じでどのバージョンの Pthon を選ぶかみたいなのが出るので、使いたいやつをクリック（多分最新版のバージョンしか選択肢に出ないと思います）



- このままエンター



- 次に仮想環境の中に入れたいパッケージがあれば選びます。numpy とか scipy, matplotlib とかとりあえず入れときましょう。




- vscode 内のエクスプローラーで今開いているフォルダを見ると「.venv」というフォルダが新しくできていると思います。これが仮想環境です。このフォルダの中にPythonパッケージを取り入れていきます。こうすることで、PC全体のグローバルな環境を汚さずに、プロジェクトごとに必要な Python パッケージを切り分けることができます。
- 今後いろいろな技術ブログや本などで見ることになる「pip install ○○」というライブラリ・パッケージをインストールする操作はここで紹介したGUIによる方法で入れると、仮想環境に簡単にインストールできます。
