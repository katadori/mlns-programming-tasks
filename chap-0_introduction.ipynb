### 1.2 Python のインストール

- ググるといろいろなページが出てきますが、現代の Python 開発環境は `uv` というパッケージで管理することが多いです。
- まず `winget` か `brew` か `curl` を使って `uv` をインストールします。
- その後 `uv` を使って Python をインストールします。たとえば以下のコマンドを実行すると最新バージョンの Python がインストールされます。

```bash
uv python install
```

- このとき PC が `uv` を認識できておらずエラーが出る可能性があります。慌てずに、エラーメッセージをきちんと読むか、Gemini とかにコピペして「どういうエラー？」とかぶん投げて聞いてください。
- 大体 PC を再起動すれば解決します。
- `uv` を使った Python のインストールは以下の公式ドキュメントが分かりやすいです。
- [Installation | uv](https://docs.astral.sh/uv/getting-started/installation/)
- `uv` の公式ドキュメントは分かりやすいので Python のインストール以外でもどんなことができるのかはこまめに参照してもいいと思います。

- 非公式のページだと以下のページとかが役に立つと思います。
- [uvを使おう](https://okumuralab.org/~okumura/python/uv.html)

- `Python` をインストール後、ターミナルまたはコマンドプロンプトで以下のコマンドを実行し、バージョンが表示されるか確認します。

```bash
python3 --version
# または Windows では
python --version
```

- `Python` をインストール後インストール後、VSCode 拡張機能から「Python」「Python Environments」拡張機能をインストールします。

### 1.3 仮想環境を作る

- Pythonでは、プロジェクト（課題や研究など）ごとにパッケージのバージョンが混ざらないよう、「仮想環境」を作るのが普通です。
- ここではさっきインストールした `uv` を使って仮想環境を作ります。
- まず、適当な作業用フォルダ（例：`mywork` や `numerical_calc`）を作成し、VSCodeでそのフォルダを開いてください。
- VSCodeのメニューから「ターミナル」→「新しいターミナル」を開き、以下のコマンドを実行します。

```bash
uv init
```

- これで現在のフォルダが Python プロジェクトとして初期化されます。
- 続いて、課題で必要になるパッケージ（Jupyterや数値計算ライブラリなど）をインストールします。ターミナルで以下のコマンドを実行してください。

```bash
uv add jupyter numpy matplotlib
```

- このコマンドを実行すると、自動的に `.venv` という仮想環境が作成され、その中にパッケージが超高速でインストールされます。
- 今後いろいろな技術ブログや本などで「`pip install ○○`」というコマンドを見ると思いますが、`uv` を使う環境では代わりに「`uv add ○○`」とすることで、仮想環境にパッケージを安全にインストール・管理できます。
- `uv pip install ○○` でインストールもできますがコレは `uv` の利点を殺しているので非推奨です。避けましょう。

### 1.4簡単なコードの実行

仮想環境が作れたはずなので、そのフォルダの中で Python がきちんと実行できるかを試してみましょう。

適当に `test1.py` とかの Python ファイルを作って

```python
print("Hello, World!)
```

とかできちんとコンソールに出力されるかを見るとよいと思います。Pythonファイルを実行するには、VSCodeの右上の再生ボタン▶を押すか、ターミナルで 

```bash
uv run python test1.py 
```

と実行します。

Jupyter notebook を動かすにはさらに拡張機能が必要なので、適当に `test2.ipynb` とかで notebook ファイルを作って

````python
a = 10
b = 5
a + b
````

を実行させてみてください。

このとき、VSCodeの上のほうで「カーネルの選択（Select Kernel）」が求められることがあります。その場合は、「Python Environments」などを選び、先ほど作成した `.venv` の中にある Python を選択してください。

また、足りない拡張機能があればVSCodeに「この拡張機能を入れてね」みたいなことを vscode に言われると思うので、その指示通りに拡張機能を入れれば jupyter notebook を動かせるようになると思います。

#### Q. なぜ uv（仮想環境）を使うのか？

→ プロジェクトごとにライブラリやPythonのバージョンを管理するため<br>

- Pythonを長く使っていると「Aの課題では古いバージョンのライブラリが必要だけど、Bの研究では最新が必要」といった依存関係の地獄が発生します。
- `uv` はそういう仮想環境の作成とパッケージ管理を非常に簡単かつ爆速で行ってくれる現代的ツールです。
- とりあえず最初は適当に一つ環境を作って (myworkとか) 後から細かいことは考えればいいと思います。
- 気になったら「Python uv とは」「仮想環境 メリット」とかで調べればいろんな記事が出てくると思います。
