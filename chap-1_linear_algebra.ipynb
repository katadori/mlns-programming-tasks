## 行列の四則演算

### 行列の和と差

ここから行列の計算の操作に入ります．
ここで
$$
    B =
    \begin{pmatrix}
        1 & -2 & 8 \\
        2 & 5 & -1
    \end{pmatrix},
$$

$$
    C =
    \begin{pmatrix}
        -2 & 5 & 1 \\
        3 & -1 & 2
    \end{pmatrix}
$$
とします．

まずB, Cを定義し，つづいて$D = B + C$と$E = B - C$ を計算してみましょう．


````python
B = [[1, -2, 8], [2, 5, -1]]

C = [[-2, 5, 1], [3, -1, 2]]
````

````python
# B + C

D = B + C
print(D)

# B - C

E = B - C

print(E)
````

$B + C$ は計算しましたが，結果は4*3の行列になってしまい，$B - C$ はそもそも計算ができませんでした．

行列の和，差は各成分ごとの和，差でした．

各成分ごとの計算を，`for`　ループを使って書いてみましょう．
Dのi,j成分は，Bのi,j成分とCのi,j成分の和になるので．．．

計算の都合上，2*3 で要素が0の行列$D, E$を先に定義しておきます．

````python
# Define D and E
D =  [[0 for j in range(3)] for i in range(2)]
E =  [[0 for j in range(3)] for i in range(2)]

# B + C
for i in range(2):
    for j in range(3):
        D[i][j] = B[i][j] + C[i][j]

print(D)

# B - C
for i in range(2):
    for j in range(3):
        E[i][j] = B[i][j] - C[i][j]

print(E)
````

### 行列のスカラー倍

次は行列のスカラー倍を計算してみましょう．

$B$ を 3倍した行列，$F=3B$を計算してみてください．
ここまでやってくれば，`3 * B` ではないことはわかりますね．

````python
# Define F
F =  [[0 for j in range(3)] for i in range(2)]

# F = 3 * B
for i in range(2):
    for j in range(3):
        F[i][j] = 3 * B[i][j]

print(F)
````

### 行列の積

ここまでくれば行列の積の計算をforループを使って計算する方法はわかるでしょう．

行列 $G, H$を次のように定義します．
$$
    G = 
    \begin{pmatrix}
        2 & 1 & -3 \\
        1 & -5 & 2
    \end{pmatrix},
$$

$$
    H = 
    \begin{pmatrix}
        3 & 1 & 0 \\
        2 & 0 & -1 \\
        -1 & 4 & 1
    \end{pmatrix}.
$$
$J = GH$ を計算してみてください．

````python
# Define J
J =  [[0 for j in range(3)] for i in range(2)]

# Define G, H

G = [[2, 1, -3], [1, -5, 2]]
H = [[3, 1, 0], [2, 0, -1], [-1, 4, 1]]

# J = G H

for i in range(2):
    for j in range(3):
        for k in range(3):
            J[i][j] += G[i][k] * H[k][j]
print(J)
````

## 3*3 行列式

3*3行列の行列式は，サラスの方法を用いて簡単に計算できますね．

以下で定められる行列 $K$:
$$
    K = 
    \begin{pmatrix}
        1 & 2 & 3 \\
        0 & 5 & 2 \\
        7 & 1 & 6
    \end{pmatrix},
$$
の行列式を計算してください．

````python
# define K
K = [[1, 2, 3], [0, 5, 2], [7, 1, 6]]

# calculate the determinant of K
det = 0


KK = [K[i]+K[i] for i in range(3)]
print(KK)

for j in range(3):
    det += KK[0][j] * KK[1][j+1] * KK[2][j+2]
    det -= KK[0][j+3] * KK[1][j+2] * KK[2][j+1]

print(det)
````

## 関数を活用する（処理の一般化）

ここまで簡単な行列の計算を実装してきました．

ここからは，さらに複雑な計算を行うために強力な2つの道具，**関数**と**ライブラリ**について学びます．

まずは，関数から見ていきましょう．

例えば，それぞれ3*3行列である，$L, M$ について, その積，$LM$ と $ML$ を計算したいとします．

この計算は，上で見たように for 文を使って書くことができますが，
ほとんど同じ計算を何度も書くことは無駄な労力です．

従って，ある $3*3$ 行列 $A$ と $B$ の積を求めるプログラムを書いて，$A, B$に任意の $3*3$行列を代入することで，任意の行列について積を求めるプログラムを書きたいと考えます．これを実現するのが関数です．

例えば関数は次のように書くことができます．

```
def new_func(a, b):
    c = a * B
     return c
```

詳しくは python のチュートリアル等を見てください．

注1) 
- 関数にわたす値や変数（上の例でいえば `a, b`）のことを「引数（**ひきすう**）」と呼びます．

注2)
- 関数の出力（上の例で言えば `return`の後の値`c`）を「戻り値」あるいは「返り値」と呼びます．

演習として，$3*3$ の行列 $A, B$ について積を求める関数を作り，以下の行列
$$
    L = 
    \begin{pmatrix}
        6 & 1 & 2 \\
        2 & 7 & -1 \\
        -1 & 4 & 1
    \end{pmatrix},
$$

$$
    M = 
    \begin{pmatrix}
        1 & -2 & 3 \\
        8 & 7 & -9 \\
        -4 & -5 & 6
    \end{pmatrix},
$$
の積，$LM$と$ML$を求めてください．


````python
L = [[6, 1, 2], [2, 7, -1], [-1, 4, 1]]
M = [[1, -2, 3], [8, 7, -9], [-4, -5, 6]]
````

````python
def calc_squarematrix33_product(mat_A, mat_B):
    mat_C =  [[0 for j in range(3)] for i in range(3)]
    for i in range(3):
        for j in range(3):
            for k in range(3):
                mat_C[i][j] += mat_A[i][k] * mat_B[k][j]    
    return mat_C

print(calc_squarematrix33_product(L, M))
print(calc_squarematrix33_product(M, L))
````

行列の積が非可換であることが確認できたでしょうか？

このプログラムだと，$3*3$ 行列の積にしか適用できません．
$4, 4$ 行列同士の積を計算したい場合，また新しくプログラムを書く必要ができてしまいます．

任意の正方行列，$A, B$ の積を計算する関数はどのように書けばよいでしょうか...?

*ヒント
行列の大きさを取得するにはどうすればいいでしょうか...?

````python
def calc_squarematrixnn_product(mat_A, mat_B):
    size = len(mat_A)
    mat_C =  [[0 for j in range(size)] for i in range(size)]
    for i in range(size):
        for j in range(size):
            for k in range(size):
                mat_C[i][j] += mat_A[i][k] * mat_B[k][j]    
    return mat_C

calc_squarematrixnn_product(L, M)
````

ここまでくると，任意の（積が定義できる）行列の積を計算するプログラムが書きたくなってくるのではないでしょうか？

一つの関数を用いて，$LM, GH$ を計算してみてください．

````python
def calc_matrix_product(mat_A, mat_B):
    size_A = len(mat_A)
    size_B = len(mat_B[0])
    size = len(mat_A[0])
    mat_C =  [[0 for j in range(size_B)] for i in range(size_A)]
    for i in range(size_A):
        for j in range(size_B):
            for k in range(size):
                mat_C[i][j] += mat_A[i][k] * mat_B[k][j]    
    return mat_C

print(calc_matrix_product(L, M))
print(calc_matrix_product(G, H))
````

このように，関数を用いることで同じ処理を何度も書く必要がなくなりました．

同じことは何度も書かない，というのはプログラムの思想として非常に重要な考え方です．

## 行列の固有値

以下，簡単のため行列は正方行列であるとします．

行列 $A$，ベクトル $\bm{v}$，定数 $\lambda$について，
$$
    A \bm{v} = \lambda \bm{v}
$$
を満たすとき，ベクトル $\bm{v}$ を行列 $A$の固有ベクトルと呼び，定数 $\lambda$ を行列 $A$ の固有値と呼びます．

### 2*2 行列の固有値

$A$ を大きさ2の正方行列，$I$ を大きさ2の単位行列，$t$ を実数とします．

このとき，固有多項式 $g_A(t)$:
$$
    g_A(t) = \mathrm{det}(tI-A),
$$
として，$g_A(t)=0$ の根が行列 $A$ の固有値になります．

例えば，
$$
    A =
    \begin{pmatrix}
        7 & -6 \\
        3 & -2
    \end{pmatrix},
$$
とすると，$A$ の固有多項式は
$$
    g_A(t) = \mathrm{det}(tI-A) =
    \left|
    \begin{matrix}
        t-7 & 6 \\
        -3 & t+2
    \end{matrix}
    \right|
    = t^2 -5t + 4 = (t-1)(t-4)
$$
となり，$A$ の固有値は $\lambda =1, 4$ である．

行列の固有値，固有ベクトルという概念は，量子力学から機械学習まで......

さまざまな場面で登場するので，固有値・固有ベクトルとは何か，ぜひ復習されたし．

ここでは, numpyを用いて固有値，固有ベクトルを計算する．

【演習】

上記の行列 $A$ の固有値，固有ベクトルをnumpyで計算する方法を調べる．

````python
arr_A = np.array([[7, -6], [3, -2]])
````

````python
u, v = np.linalg.eig(arr_A)
````

````python
print('固有値は\n{}\n'.format(u))

print('固有ベクトルは\n{}'.format(v))
````

## 行列の対角化

固有値，固有ベクトルと併せて，さまざまな場面でよく出てくる数学の道具が対角化です．

行列 $A$ が，その成分が実数である対称行列であるとき，
$$
    P^{-1} A P = \Lambda
$$
となる行列 $P$ が存在する．
ただし，$P$ は $P^T = P^{-1}$ を満たす直交行列であり，$\Lambda$ は対角行列である．

このとき，$P$ は $A$ の固有ベクトル（列ベクトル）を横に並べた行列であり，
$\Lambda$ の対角成分はそれぞれ対応する固有値であることが知られている．

【演習】
前節で用いた方法で固有ベクトルを求め，直交行列 $P$ を求める．
これによって $\Lambda$ を計算し，それが対角行列になっていることと，その成分が固有値に対応していることを確かめる．

````python
arr_A = np.array([[7, -6], [-6, -2]])

la, P = np.linalg.eig(arr_A)

print('固有値は\n{}\n'.format(la))

print('Pは直交行列である: \n P P^T=\n{}\n'.format(P @ P.T))

L = P.T @ arr_A @ P

print('行列 L\n{}'.format(L))
````

行列 $A$ が対称行列でなくても，対角化可能な場合については，読者の課題とする．

応用面において「実対称行列が直交行列で対角化可能」であることを用いる場面がしばしば存在するので，ここでは実対称行列のみを扱った．

行列 $A$ が正方行列でない場合も，対角化と類似の計算が存在する→特異値分解
